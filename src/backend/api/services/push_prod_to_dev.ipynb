```python
from sql_connection import read_table, write_table, read_query
import pandas as pd

est_raw_df =  pd.read_excel(r"/home/e10dz3/Project/EH%20Vehicle%20Repair/data_api/api_est_raw.xlsx")
est_line_df = pd.read_excel(r"/home/e10dz3/Project/EH%20Vehicle%20Repair/data_api/api_est_line.xlsx")
subtot_df = pd.read_excel(r"/home/e10dz3/Project/EH%20Vehicle%20Repair/data_api/api_est_subtot.xlsx")
```

```python
est_raw_df['est_id'] = est_raw_df['est_id'].astype(int).astype(str)
est_line_df['est_id'] = est_line_df['est_id'].astype(int).astype(str)
subtot_df['est_id'] = subtot_df['est_id'].astype(int).astype(str)
```

```python
import numpy as np

cols = ['cieca_dtl_hdr_id', 'cieca_part_dtl_line_id', 'cieca_lbr_dtl_line_id', 'cieca_othr_chrg_dtl_line_id']

est_line_df[cols] = est_line_df[cols].astype('Int64').astype(str).replace("<NA>", np.nan).replace("nan", np.nan)
subtot_df['adj_typ_id'] = subtot_df['adj_typ_id'].astype('Int64').astype(str).replace("nan", np.nan).replace("<NA>", np.nan)
```

```python
import numpy as np
_NULL_STRINGS = {"none", "null", "na", "n/a", "nan", ""}
def _sanitise(df: pd.DataFrame) -> pd.DataFrame:
    """Replace null-like strings with NaN so Postgres float columns don't choke."""
    null_strings = {s for s in _NULL_STRINGS} | {s.upper() for s in _NULL_STRINGS}
    return df.where(~df.isin(null_strings), other=np.nan)

_sanitise(est_raw_df)
_sanitise(subtot_df)
_sanitise(est_line_df)
```

```python
write_table(est_raw_df, table_name= "m4_api_est_imgs", schema= "container", if_exists="replace")
write_table(subtot_df, table_name= "m4_api_est_subtot", schema= "container", if_exists="replace")
write_table(est_line_df, table_name= "m4_api_est_line", schema= "container", if_exists="replace")
```
